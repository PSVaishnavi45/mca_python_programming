# map()

```python
map(int,['1','3','5','10'])
```

```python
#simple extraction from map using loop
for num in map(int,['1','3','5','10']):
    print(num)
```

```python
for ascii_code in map(ord,['A','B','C','D']):
    print(ascii_code,end=" ")
```

```python
animals=['horse','rabbit','deer','fox','dog','tigger','lion','wolf']
out=list(map(lambda e: e[0],animals))
print(out)
```

```python
list(map(lambda e: e[:3],animals))
```

```python
fruits=['apple','orange','banana','berry','cherry','chestnut','guava','mango']
print(fruits)
print(len(fruits)==len(animals))
```

```python
 #first 3 letters of animals and last 3 letter of fruits
list(map(lambda a,f:a[:3] + f[-3:],animals,fruits))
```

```python
x,y=[1,2,3],[3,4,5]
z=list(map(lambda i,j: i+j,x,y))
print(x,y,z,sep='\n')
```

# filter()
filter takes boolean function

```python
x=[1,-2,-3,4,5,6]
ans=filter(lambda i: i>0,x)
list(ans)
```

```python
list(filter(lambda name:'a' in name,animals))
```

```python
list(filter(lambda name:'a' not in name,animals))
```

```python
list(filter(lambda e:len(e)>5,animals))
```

```python
class Student: 
    def __init__(self, name, marks=[]):
        self.name = name
        self.marks = marks
    def total_marks(self): return sum(self.marks)
    def __gt__(self, other):
        if isinstance(other, Student):
            return self.total_marks() >other.total_marks()
        elif isinstance(other, int):
            return self.total_marks() >other
            
    def __repr__(self):
        return f'Student({self.name},{self.total_marks()})'

#outside the function
def is_passed(marks):
        return marks > 200
students = [
    Student('shami',[23, 56, 67, 20, 10]),
    Student('neha',[43, 86, 77]),
    Student('vaishu',[83, 86, 87]),
    Student('kaavya',[53, 86, 67]),
]

list(filter(is_passed, students))
```
