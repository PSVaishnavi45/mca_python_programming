```python
# no constructor
class Area:
    leng=0
    bth=0
    def calc(self):     #self->it tells python what item it belongs to which class
        ans=self.leng*self.bth
        return
```

```python
# constructor
class Area2:
    def __init__(self,leng,bth):  #constructor      magic methods/dundur methods-> method that strart with double underscore and ends with same
        self.l=leng
        self.b=bth
    def calc(self):        # method
        ans=self.l*self.b
        return ans
```

```python
o1=Area()
o1.leng=5
o1.bth=9
ans=o1.calc()
print(o1.leng,o1.bth,ans)

```

```python
class Pokemon:
    def __init__(self,name,type='normal',hp=10,atk=10,dfn=10):
        self.name=name
        self.type=type
        self.hp=hp
        self.atk=atk
        self.dfn=dfn
    def states(self):
        print("pokemon details")
        print(f'Name:{self.name}')
        print(f'Type:{self.type}')
        print(f'hp:{self.hp}')
        print(f'ATK:{self.atk}')
        print(f'DFN:{self.dfn}')
    
    def overall(self):
        return self.hp+self.atk+self.dfn
    
    def is_powerfull(self):
        if self.overall()>100:
            return True
        return False
```

```python
pikachu=Pokemon("pikachu","electric",35,55,45)
snorlax=Pokemon("snorlax","normal",35,32,67)
charizard=Pokemon("charizard","fire",12,45,35)
bulbasaur=Pokemon("bulbasaur","Grass",35,25,45)
ratata=Pokemon("ratata")
pokemons=[pikachu,snorlax,charizard,bulbasaur,ratata]
print("\nstates")
for poke in pokemons:
    poke.states()
    print(f'overall score:{poke.overall()}')
    print(f'is powerfull:{poke.is_powerfull()}')
    print('---------------')
    
```

```python
class Calc:
    def run(self,num1,operator,num2): 
        match operator:
            case'+':return self.add(num1,num2)
            case'-':return self.subtract(num1,num2)
            case'/':return self.divide(num1,num2)
            case'*':return self.multiply(num1,num2)
    def add(self,a,b): return a+b
    def multiply(self,a,b): return a*b
    def substract(self,a,b): return a-b
    def divide(self,a,b):
        if b==0: return "can't divide by 0"
        return a/b
    
                
```

```python
#calculator instances
c=Calc()
while True:
    a=int(input("enter first num: "))
    b=int(input("enter next num: "))
    op=input("enter operation(+,-,/,*)")
    ans=c.run(a,op,b)
    print(ans)
    if input('continue')!='y':
        break
```

# method overriding

```python
class Vehicle:
    def __init__(self,brand):
        self.brand=brand
    def start(self):
        print("vehicle started")
    def stop(self):
        print("vehicle stopped")
```

```python
class Car(Vehicle):
    def start(self):                         #method overridden
        print("the car is started")
    def stop(self):                          #method overridden
        print("the car is stopped")
```

```python
vh=Vehicle('abc')
vh.start()
vh.stop()
car1=Car('honda')
car1.start()
car1.stop()
```

# magic/dundur methods

magic methods/dundur methods-> method that strart with double underscore and ends with same
        
>  __gt__()
<  __it__()
== __eq__()
+  __add__()
-  __sub__()
   __str__()#string representation
   __repr__()

```python
class Player:
    def __init__(self,name,score):
        self.name=name
        self.score=score
    def __gt__(self,other):
        if isinstance(other,Player):# funtion used to validate if an object belongs to a specific class or not
            return self.score>other.score
```

```python
p4=Player('kaladin',70)
p5=Player('rohit',80)
if p4>p5:
    print(f"{p4.name} is better")
else:
    print(f"{p5.name} is better")
```

```python
class Student:
    def __init__(self,name,roll,gender):
        self.name=name
        self.roll=roll
        self.gender=gender
    def __str__(self):#transforms string into an objects
        return f'Student({self.name})'
    def __repr__(self):
        return f'Student({self.name})'
        
```

```python
s1=Student('vaishu','2s3','Female')
s2=Student('neha','2s5','Female')
s3=Student('bindu','5r9','male')
s4=Student('hari','1a8','male')
print(s1)
students=[s1,s2,s3,s4]
print(students)
```
